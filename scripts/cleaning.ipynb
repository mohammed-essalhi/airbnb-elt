```python
import pandas as pd
```

```python
# load the dataset
df = pd.read_csv('../sample_data/airbnb_sample.csv', low_memory=False)
df
```

```python
# return the columns of the dataframe
df.columns
```

```python
# return the number of null values in each column
df.isna().sum()
```

```python
# return unique values in the 'country' column
df['country'].unique()
```

```python
# return unique values in the 'country_code' column
df['country_code'].unique()
```

```python
# drop the columns that are not needed for our analysis

columns_to_drop = [ 'country', 'country_code', # we droping this because it is redundant, and have only one value
                    'id',   # we droping this because we have host_id as unique identifier
                    'last_review', 'reviews_per_month', # we droping these because they're not useful for our analysis
                    'license',  # we droping this because it is all empty values
                ] 
df.drop(columns=columns_to_drop, inplace=True)
```

```python
df.shape
```

```python
# Replace null values in 'house_rules' with 'No rules'
df['house_rules'] = df['house_rules'].fillna('No rules')
```

```python
df.isna().sum()
```

```python
# delete rows with null values
df.dropna(inplace=True)
```

```python
df.isna().sum()
```

```python
df.shape
```

```python
df['host_identity_verified'].unique()  
```

```python
# simply get 1 if it can be booked instantly, 0 otherwise
df['instant_bookable'] = df['instant_bookable'].apply(lambda x: 1 if x == True else 0)
```

```python
# if the host is verified, we get 1, otherwise 0
df['host_identity_verified'] = df['host_identity_verified'].apply(lambda x: 1 if x == "verified" else 0)
```

```python
# Reset the index to a new sequential index
df = df.reset_index(drop=True)
df.head()
```

```python
# check the type of the first element in the 'price' column
type(df['price'][0])
```

```python
# price &service_fee are strings, so we need to convert them to float while removing the dollar sign and commas
df['service_fee'] = df['service_fee'].str.replace('$', '').str.replace(',', '').astype(float)
df['price'] = df['price'].str.replace('$', '').str.replace(',', '').astype(float)
df.head()
```

```python
# Convert all column names to lowercase
df.columns = df.columns.str.lower()

# Verify the new column names
df.columns
```

```python
# change the type of these columns to int
df['construction_year'] = df['construction_year'].astype(int)
df['number_of_reviews'] = df['number_of_reviews'].astype(int)
df['minimum_nights'] = df['minimum_nights'].astype(int)
df['availability_365'] = df['availability_365'].astype(int)
df.info()
```

```python
# check how many values are out of range in the 'availability_365' column
df[(df['availability_365'] < 0) | (df['availability_365'] > 365)].shape[0]
```

```python
# remove any values that are out of range in the 'availability_365' column
import numpy as np
df['availability_365'] = df['availability_365'].clip(lower=1, upper=365)

# check how many values are out of range in the 'availability_365' column again
df[(df['availability_365'] < 0) | (df['availability_365'] > 365)].shape[0]
```

```python
# renaming columns
df.rename(columns={
    'host_identity_verified': 'is_host_verified',
    'lat': 'latitude',
    'long': 'longitude',
    'instant_bookable': 'is_instant_bookable',
    'number_of_reviews': 'review_count',
    'review_rate_number': 'review_rating',
    'calculated_host_listings_count': 'host_listings_count',
}, inplace=True)
df.head(5)
```

```python
import psycopg2
from sqlalchemy import create_engine

# Database connection parameters for first_ELT_DB
db_params = {
    'dbname': 'first_ELT_DB',
    'user': 'postgres',
    'password': '<YOUR_POSTGRES_PASSWORD>',
    'host': 'localhost',
    'port': '5432'
}

# Schema and table details
schema = 'public'
table_name = 'pd_cleaned_data'

# Create PostgreSQL table
try:
    conn = psycopg2.connect(**db_params)
    cursor = conn.cursor()
    create_table_query = f"""
    CREATE TABLE IF NOT EXISTS {schema}.{table_name} (
        name TEXT,
        host_id BIGINT,
        is_host_verified BOOLEAN,
        host_name TEXT,
        neighbourhood_group TEXT,
        neighbourhood TEXT,
        latitude NUMERIC,
        longitude NUMERIC,
        is_instant_bookable BOOLEAN,
        cancellation_policy TEXT,
        room_type TEXT,
        construction_year INTEGER,
        price NUMERIC,
        service_fee NUMERIC,
        minimum_nights INTEGER,
        review_count INTEGER,
        review_rating NUMERIC,
        host_listings_count NUMERIC,
        availability_365 INTEGER,
        house_rules TEXT
    );
    """
    cursor.execute(create_table_query)
    conn.commit()
    print(f"Table {schema}.{table_name} created or already exists.")
    cursor.close()
except Exception as e:
    print(f"Error creating table: {e}")
    conn.rollback()
finally:
    conn.close()

# Load DataFrame to PostgreSQL
try:
    # Convert is_host_verified and is_instant_bookable to boolean for PostgreSQL
    df['is_host_verified'] = df['is_host_verified'].map({1: True, 0: False})
    df['is_instant_bookable'] = df['is_instant_bookable'].map({1: True, 0: False})

    # Create SQLAlchemy engine
    conn_string = f"postgresql+psycopg2://{db_params['user']}:{db_params['password']}@{db_params['host']}:{db_params['port']}/{db_params['dbname']}"
    engine = create_engine(conn_string)

    # Load df to PostgreSQL
    df.to_sql(table_name, engine, schema=schema, if_exists='replace', index=False, chunksize=10000)
    print(f"Data loaded successfully into {schema}.{table_name} with {len(df)} rows.")

except Exception as e:
    print(f"Error loading data to PostgreSQL: {e}")
```
